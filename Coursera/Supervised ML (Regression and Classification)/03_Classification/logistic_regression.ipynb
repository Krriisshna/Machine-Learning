# Logistic Regression

## Sigmoid or Logistic Function
- In Linear Regression model f<sub>w,b</sub>(x<sup>(i)</sup>) = w&middot;x<sup>(i)</sup> + b to predict y given x.
- But, in classification we want prediction to be between 0 and 1.
- Since, our output y is either 1 or 0.
- This can be accomplished by sigmoid function which will map all input values between 0 and 1.
### Formula for sigmoid function:
- g(z) = $\frac{1}{1 + e^{-z}}$  
in the case of logistic regression, z(the input of logistic regression), is the output of a linear regression model.


```python
%matplotlib widget
import matplotlib.pyplot as plt
import numpy as np
from function import draw_vthresh
```

```python
input_array = np.array([1,2,3])
exp_array = np.exp(input_array)
# np.exp() is NumPy's exponential function use to calculate eᶻ

print("input to exp:", input_array)
print("output of exp:", exp_array)

input_val = 1
exp_val = np.exp(input_val)

print("input to exp:", input_val)
print("output of exp:", exp_val)
```

```python
def sigmoid(z):
    g = (1)/(1 + np.exp(-z))
    return g
```

```python
z_tmp = np.arange(-10,11)

y = sigmoid(z_tmp)

np.set_printoptions(precision=3)
# this will display 3 decimal places

print(" Input(z),    Output (sigmoid(z))")
print(np.c_[z_tmp, y])
```

```python
fix, ax = plt.subplots(1,1, figsize=(8,5))
ax.plot(z_tmp,y, color='b')

ax.set_title("Sigmoid Function")
ax.set_ylabel("sigmoid(z)")
ax.set_xlabel("z")
draw_vthresh(ax,0)
plt.show()
```
