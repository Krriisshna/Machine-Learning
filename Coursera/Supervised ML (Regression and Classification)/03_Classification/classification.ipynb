# Classification
- Classification Problems:
Examples of classification problems are things like: identifying email as Spam or Not Spam or determining if a tumor is malignant or benign. In particular, these are examples of binary classification where there are two possible outcomes.

```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
def plot_data(X,y,ax, pos_label="y=1",neg_label="y=0",s=80,loc="best"):

    pos = y_train == 1
    neg = y_train == 0
    pos.reshape(-1,)
    neg.reshape(-1,)

    ax.scatter(X[pos,0],X[pos,1],marker='x',s=s,c='red',label=pos_label)
    ax.scatter(X[neg,0],X[neg,1],marker='o',s=s,label=neg_label,facecolors='none',edgecolors='blue',lw=3)
    # s is area of marker, facecolors decides hollowness.
    # edgecolors outline color, lw is width of outline
    ax.legend(loc=loc)
    # loc='best' means best position for legend within plot.
```

```python
x_train = np.array([0., 1, 2, 3, 4, 5])
y_train = np.array([0,  0, 0, 1, 1, 1])
x_train2 = np.array([[0.5, 1.5], [1,1], [1.5, 0.5], [3, 0.5], [2, 2], [1, 2.5]])
y_train2 = np.array([0, 0, 0, 1, 1, 1])
```

```python
pos = y_train == 1
neg = y_train == 0

fig, ax = plt.subplots(1,2, figsize=(8,3))

#plot 1, single variable
ax[0].scatter(x_train[pos],y_train[pos],marker='x',s=80,c='red',label='y=1')
ax[0].scatter(x_train[neg],y_train[neg],marker='o',s=100,facecolors='none',edgecolors='blue',label='y=0',
              lw=3)
ax[0].set_ylim(-0.08,1.1)
# set_ylim sets y axis display limits

ax[0].set_xlabel('x',fontsize=12)
ax[0].set_ylabel('y',fontsize=12)
ax[0].set_title('one variable plot')
ax[0].legend()

#plot 2, two variables
plot_data(x_train2,y_train2,ax[1])
ax[1].axis([0,4,0,4])
# .axis([x_min,x_max,y_min,y_max]) it sets display limits of both axis.

ax[1].set_xlabel('$x_0$',fontsize=12)
ax[1].set_ylabel('$x_1$',fontsize=12)
ax[1].set_title('two varaibles plot')
ax[1].legend()
```
