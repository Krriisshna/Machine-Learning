## Logistic Loss function
**Loss** is a measure of the diffrence of a single example to its target value.  
**Cost** is a measure of the losses over the training set.  

- loss($f_w,_b$($x^{(i)}$),$y^{(i)}$) is the cost for a single point, which is:  
    
- $f_{w,b}$($x^{(i)}$) is model's prediction, while $y^{(i)}$ is target value.
- $f_{w,b}$($x^{(i)}$) = g(w.$x^{(i)}$ + b), where g is sigmoid function.

logistic-regression cost curve may not look like a simple bowl, but it is still easy for gradient descent to optimize.
- No local minima: gradient descent will not get stuck at a wrong low point.
- No plateaus: the slope does not become completely flat before reaching the minimum.
- No discontinuities: the curve is smooth, so gradients are well-defined.
