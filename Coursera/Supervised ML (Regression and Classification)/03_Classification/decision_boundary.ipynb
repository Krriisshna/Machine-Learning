# Logistic Regression, Decision Boundary

```python
import numpy as np
%matplotlib widget
import matplotlib.pyplot as plt
from function import sigmoid,plot_data,draw_vthresh
```

```python
# Data
X = np.array([[0.5, 1.5], [1,1], [1.5, 0.5], [3, 0.5], [2, 2], [1, 2.5]])
y = np.array([0,0,0,1,1,1]).reshape(-1,1)
```

```python
# Plot Data
fig,ax = plt.subplots(1,1, figsize=(4,4))
plot_data(X,y,ax)

ax.axis([0,4,0,3.5])
ax.set_xlabel('$x_0$')
ax.set_ylabel('$x_1$')
plt.show()
```

## Logistic Regression Model
- Let form of model  
  f(x) = g($w_0x_0 + w_1x_1$ + b)
  where, g(z) = $\frac{1}{1 + e^{-z}}$
- Let we got parameters as b = -3, $w_0 = 1, w_1 = 1$ then,  
  f(x) = g($x_0 + x_1 -3$)

```python
# Plot sigmoid(z)
z = np.arange(-10,11)
fig,ax = plt.subplots(1,1, figsize=(5,3))

ax.plot(z,sigmoid(z),color='blue')
ax.set_title("Sigmoid function")
ax.set_ylabel("sigmoid(z)")
ax.set_xlabel("x")
draw_vthresh(ax,0)
```

## Plotting Decision Boundary
- Our Logistic Regression model is:  
  f(x) = g($x_0 + x_1 - 3$)
- it predicts y = 1 if $x_0 + x_1 - 3 = 0$
- so we will plot $x_0 + x_1 -3 = 0$ which is equivalent to $x_1 =  3 - x_0$

```python
x0 = np.arange(0,6)
x1 = 3 - x0

fig,ax = plt.subplots(1,1, figsize=(5,4))

# Decision Boundary plot
ax.plot(x0,x1,c="b")
ax.axis([0,4,0,3.5])

ax.fill_between(x0,x1, alpha=0.2)
# .fill_between shades the area curve
# alpha sets the transparency

# Original Data plot
plot_data(X,y,ax)
ax.set_ylabel(r'$x_1$')
ax.set_xlabel(r'$x_0$')
plt.show()
```
