## Cost Function
For logistic regression, the cost function is:  
    
where,  
- **loss($f_{w,b}(x^{(i)}),y^{(i)}$)**, is the cost for a single training example.  
  loss($f_{w,b}(x^{(i)}),y^{(i)}$) = $-y^{(i)}log(f_{w,b}(x^{(i)}))-(1-y^{(i)})log(1-f_{w,b}(x^{(i)}))$
- m is the number of training examples.
- $f_{w,b}(x^{(i)}) = g(z^{(i)})$

```python
def compute_cost_logistic(X,y,w,b):
    m = X.shape[0]
    cost = 0.0

    for i in range(m):
        z_i = np.dot(X[i],w) + b
        f_wb_i = sigmoid(z_i)
        cost += -y[i]*np.log(f_wb_i) - (1-y[i])*np.log(1 - f_wb_i)

    cost = cost/m
    return cost
```

```python
w_tmp = np.array([1,1])
b_tmp = -3
print(compute_cost_logistic(X_train,y_train,w_tmp,b_tmp))
```

## Plot Decision Boundary
- plotting decision boundary for two different values of b
- for b = -3,$w_0 = 1$,$w_1 = 1$, we'll plot (-3 + $x_0 + x_1 = 0$)
- for b = -4,$w_0 = 1$,$w_1 = 1$, we'll plot (-4 + $x_0 + x_1 = 0$)

```python
x0 = np.arange(0,6)

x1 = 3 - x0
x1_other = 4 - x0

fix,ax = plt.subplots(1,1, figsize=(4,4))

# plotting decision boundary
ax.plot(x0,x1, c=dlc['dlblue'],label='b=-3')
ax.plot(x0,x1_other,c=dlc['dlmagenta'],label='b=-4')
ax.axis([0,4,0,4])

# plotting the original data
plot_data(X_train,y_train,ax)
ax.axis([0,4,0,4])
ax.set_ylabel("$x_1$",fontsize=12)
ax.set_xlabel("$x_0",fontsize=12)
ax.legend(loc='upper right')
plt.title('Decision Boundary')
plt.show()
```

- from this plot b = -4 is worse, lets see cost function implementation reflects this.

```python
w_array1 = np.array([1,1])
b_1 = -3
w_array2 = np.array([1,1])
b_2 = -4

print("cost for b = -3: ",compute_cost_logistic(X_train,y_train,w_array1,b_1))
print("cost for b = -4: ",compute_cost_logistic(X_train,y_train,w_array2,b_2))
```

- by cost function implementation b = -4 is giving higher cost than b = -3.
